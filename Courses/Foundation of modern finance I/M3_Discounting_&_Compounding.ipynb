Q1:

```python
# 10 annual payments of $2300 to payoff loan. Make first payment a year from now
# Bank offers APR 5%

# Amount to deposit today?
paymentAmount = 2300
#total = paymentAmount
time = 10
bank_apr = 5/100
# Using present value annuity
display(Math(r'PV_{annuity} = A \frac{1}{r}* \left(1 - \frac{1}{(1 + r)^t}\right)'))
print('\n')

pv_0 = (paymentAmount/bank_apr)*(1 - (1/(1 + bank_apr)**time))
f'The Present value of the amount that will be paid in ten years is ${pv_0:.4f}'
```

```python
# create table that demostrates yearly payments is enough.
balance = pv_0
amount = 2300
rows = []
for year in range(1, 10 + 1, 1):

  payment = amount - balance*bank_apr
  interest = balance*bank_apr
  endYearBalance = balance - payment
  balance = endYearBalance

  #print(f'In Year {year}, the interest {interest:.2f}, and End-of-Year Balance {endYearBalance:.2f}')
  rows.append({"Year": year, "Interest": round(interest, 2), "End-of-Year Balance": round(endYearBalance, 2)})

df = pd.DataFrame(rows)
df
```

Q3:

```python
# Patent will last 15 years.
# Profits of 1M in Year 1
# From t=1 - t = 15 growth rate is 4% per year.
# Discount rate is 9%

# What is the present value?
profit = 5000000
growthR, discountR = 4/100, 9/100
time = 15

pv_now = profit/(discountR - growthR)*(1 - ((1 + growthR)/(1 + discountR))**time)
f'The present value is ${pv_now:.2f}'
```

```python
# Rental Property.  Discount rate 7%
# Scenario 1: rent it out immediately. Expect rental profit to be $650K and to continue in perpetuity. first payment in year 1
# Scenario 2: redevelop it (takes four years). PV of redevelopment cost $1M. at t = 5, rental profit of $950K. Will continue in perpetuity.

# Scenario 1:
discountR = 7/100
pv_rentNow = 650000/discountR
print(f'The present value of the property is today is ${pv_rentNow:.2f}\n')

# Scenario 2:
pv_rentYear5 = (1/((1 + discountR)**4))*(950000/discountR)
print(f'The value of the property in year 5 is ${pv_rentYear5:.2f}\n')

# subtract cost
pvAfterCost = pv_rentYear5 - 1000000
print(f'Taking cost into account ${pvAfterCost:.2f}')
```
