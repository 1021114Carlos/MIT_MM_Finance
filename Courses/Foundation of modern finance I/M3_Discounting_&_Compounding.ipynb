```python
# Using present value annuity
display(Math(r'PV_{annuity} = A \frac{1}{r}* \left(1 - \frac{1}{(1 + r)^t}\right)'))
```

```python
payment = 2300
bank_apr = 5/100
years = 10
pv0 = (payment/bank_apr)*(1 - (1/(1 + bank_apr)**years))
balance = pv0
rows = []

for year in range(1, years + 1):
  interest = balance*bank_apr
  end_balance = balance + interest - payment

  rows.append({"Beginning Balance": f"{balance:.4f}",
               "Interest": f"{interest:.4f}",
               "Payment": payment, "E.o.Y. Balance": f"{end_balance:.4f}"
               })

  balance = end_balance

df = pd.DataFrame(rows, index=[f"Year {i}" for i in range(1, years + 1)])
df
```

```python
# Annuity with growth
display(Math(r"Annuity \ with \ growth = \frac{1}{(r - g)}\left[1 - \left(\frac{1 + g}{1 + r}\right)^t\right] \ and \  \ \frac{t}{1 + r} \ if \ r = g"))
```

```python
# Patent will last 15 years.
# Profits of 1M in Year 1
# From t=1 - t = 15 growth rate is 4% per year.
# Discount rate is 9%

# What is the present value?
profit = 5000000
growthR, discountR = 4/100, 9/100
time = 15

pv_now = profit/(discountR - growthR)*(1 - ((1 + growthR)/(1 + discountR))**time)
f'The present value is ${pv_now:,.4f}'
```

```python
# Rental Property.  Discount rate 7%
# Scenario 1: rent it out immediately. Expect rental profit to be $650K and to continue in perpetuity. first payment in year 1
# Scenario 2: redevelop it (takes four years). PV of redevelopment cost $1M. at t = 5, rental profit of $950K. Will continue in perpetuity.

# Scenario 1:
discountR = 7/100
pv_rentNow = 650000/discountR
print(f'The present value of the property today is ${pv_rentNow:,.2f}\n')

# Scenario 2:
pv_rentYear5 = (1/((1 + discountR)**4))*(950000/discountR)
print(f'The value of the property in year 5 is ${pv_rentYear5:,.2f}\n')

# subtract cost
pvAfterCost = pv_rentYear5 - 1000000
print(f'Taking re-development cost into account ${pvAfterCost:,.2f}')
```

```python
"""Both loans have monthly payments"""

new_apr = 4.5/100
mort_balance = 300_000
old_ear = 6/100 # Fixed
T = 12 # Time remaining on life of loan

# Finding old apr
old_apr = (1 + old_ear)**(1/T) - 1
print(f"Old apr: {old_apr*100:.4f}%\n")

# New apr monthly compounding
apr_monthly_compounding = new_apr/12
print(f"New apr: {apr_monthly_compounding*100:.4f}%")
```

```python
# old monthly payments
pv_old_monthly_payments = mort_balance/((1/old_apr)*(1 - (1/(1 + old_apr)**(T*20))))
print(f"Old payment amount: $ {pv_old_monthly_payments:.4f}\n")

# New monthly payments
pv_new_monthly_payments = mort_balance/((1/apr_monthly_compounding)*(1 - (1/(1 + apr_monthly_compounding)**(T*20))))
print(f"New payment amount: $ {pv_new_monthly_payments:.4f}")

payment_diff =  pv_old_monthly_payments - pv_new_monthly_payments
print(f"Saving Difference $ {payment_diff:.4f}")
```

(B)

```python
# Refinance the home with bigger mortgage. same old monthly payments
pv_new_mort_balance = pv_old_monthly_payments*((1/apr_monthly_compounding)*(1 - 1/((1 + apr_monthly_compounding)**(T*20))))
print(f"New mortgage balance: $ {pv_new_mort_balance:,.4f}")
```
