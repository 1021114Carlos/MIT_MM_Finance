```python
# Time and risk relation
t = np.linspace(0, 30, 100)    # time from 0 to 30 years
r_values = [0.04, 0.08, 0.12]
styles = [("black", "-"), ("red", "--"), ("blue", ":")]

# plotting
plt.figure(figsize=(10,6))

for r, (color, ls) in zip(r_values, styles):
  pv = 1/(1 + r)**t
  plt.plot(t, pv, label=f'r = {r}', color=color, linestyle=ls, linewidth=1.5)

plt.title('FV of $1 in year t', fontsize=14, pad=15)
plt.xlabel("Year when $1 is received", fontsize=14)
plt.ylabel('', fontsize=12)

# set axis limits
plt.xlim(0, 30)
plt.ylim(0, 1.1)

plt.xticks(np.arange(0, 31, 2))

plt.yticks(np.arange(0, 1.1, 0.2), [f'${val:.1f}' for val in np.arange(0, 1.1, 0.2)])

# add horizontal grid lines
plt.grid(axis='y', linestyle='-', alpha=0.7)
plt.grid(axis='x', linestyle='-', alpha=0.5)
plt.legend(loc='upper center', ncol=3, frameon=False, fontsize=11)

plt.tight_layout()
plt.show()
```

Calculate CF for business decision:

```python
def calculate_npv_with_details(name, cash_flows, initial_investment, rate):

  pvs = [cf / (1 + rate)**t for t, cf in enumerate(cash_flows, 1)]
  total_pv = sum(pvs)
  npv = total_pv - initial_investment

  for year, pv in enumerate(pvs, 1):
      print(f"Year {year} PV: {pv:,.4f}")

  print(f"Total PV: {total_pv:,.4f}")
  print(f"Net Present Value (NPV): {npv:,.4f}\n")

companies = {"Company A": {"flows": [500, 400, 300], "init": 1000},
             "Company B": {"flows": [-200, 300, 300], "init": 200}}

discount_rate = 0.05

# all companies
for name, data in companies.items():
    calculate_npv_with_details(name, data["flows"], data["init"], discount_rate)
```

```python
def calculate_expected_return(current_price, future_outcomes, probabilities):
  expected_future_price = np.dot(probabilities, future_outcomes)
  expected_return = ((expected_future_price/current_price)-1)*100
  return expected_return

states = [0.3, 0.4, 0.3]
assets = {'Bonds': {'current': 96, 'future': [100, 100, 100], 'color': '#1f77b4'},
          'Stocks': {'current': 80, 'future': [120, 75, 130], 'color': '#ff7f0e'}}

# for name, prices in assets.items():
#   ret = calculate_expected_returns(prices['current'], prices['future'],states)
#   print(f"{name} return is {ret:.4f}")

# Plotting
state_labels = ['State 1', 'State 2', 'State 3']
x = np.arange(len(state_labels))
width= 0.3
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))

for i, (name, data) in enumerate(assets.items()):
  future = np.array(data['future'])
  returns = ((future/data['current'])-1)*100
  expected_ret = calculate_expected_return(data['current'], data['future'], states)
  expected_price = np.dot(states, data['future'])

  offset = width*i - (width*(len(assets)-1)/2)

  # left plot: future prices + expected value line
  ax1.bar(x + offset, future, width, label=f"{name} Price", color=data['color'], alpha=0.6)
  ax1.axhline(expected_price, color=data['color'], linestyle='--', linewidth=2,
              label=f"Exp. {name}: ${expected_price:.1f}")

  # Right plot: percentage returns + expected return
  ax2.bar(x + offset, returns, width, label=f"{name} Return", color=data['color'], alpha=0.6)
  ax2.axhline(expected_ret, color=data['color'], linestyle='--', linewidth=2,
              label=f"Exp. {name}: {expected_ret:.1f}%")

# formatting
ax1.set_title("Future Prices vs. Expected Value", fontsize=14)
ax1.set_ylabel("Price ($)")
ax1.set_xticks(x)
ax1.set_xticklabels(state_labels)
ax1.legend(loc='upper left', fontsize='small')

ax2.set_title("State Returns vs. Expected Return", fontsize=14)
ax2.set_ylabel("Return (%)")
ax2.set_xticks(x)
ax2.set_xticklabels(state_labels)
ax2.axhline(0, color='black', linewidth=1) #0% baseline
ax2.legend(loc='upper left', fontsize='small')

plt.tight_layout()
plt.show()
```

(Q7.A) Finding values in Real and Nominal terms

```python
cfg = {'revenue_base': 650000, 'real_growth': 3/100, 'inflation': 2.5/100,
       'real_discount': 6/100, 'years': np.arange(1,6)}

rev_real_t1 = cfg['revenue_base']*(1 + cfg['real_growth'])

pv_real_t1 = rev_real_t1/(1+cfg['real_discount'])

rev_nom_t1 = rev_real_t1*(1 + cfg['inflation'])

pv_nom_t1 = pv_real_t1

print(f"{'Metric': <30} | {'Year 1 value':>15}")
print("-"*48)
print(f"{'Expected real revenue':<30} | ${rev_real_t1:14,.4f}")
print(f"{'PV of real revenue': <30} | ${pv_real_t1:14,.4f}")
print(f"{'Expected nominal revenue':<30} | ${rev_nom_t1:14,.4f}")
print(f"{'pv of nominal revenue':<30} | ${pv_nom_t1:14,.4f}")

# Generate graph (compounding over time)
real_curve =cfg['revenue_base']*(1 + cfg['real_growth'])**cfg['years']
nom_curve = real_curve*(1 + cfg['inflation'])**cfg['years']


plt.figure(figsize=(10, 5), facecolor='#f8f9fa')
plt.plot(cfg['years'], real_curve, 'o-', label='real revenue (base buying power)', color='#1f77b4')
plt.plot(cfg['years'], nom_curve, 's--', label='Nominal revenue (current dollars)', color='#2ca02c')

plt.title('Rental Revenue Projection: Real vs. Nominal', fontsize=14, pad=20)
plt.xlabel("Year", fontweight='bold')
plt.ylabel("Revenue ($)", fontweight='bold')
plt.xticks(cfg['years'])
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend(frameon=True, shadow=True)
plt.tight_layout()
plt.show()
```

Find prices and expected returns

```python
probabilities = np.array([0.30, 0.55, 0.15])
state_prices = np.array([0.40, 0.20, 0.10])
bonds = np.array([100, 100, 100])
stocks = np.array([85, 105, 125])

# Find bond and stock prices today.

bond_price_today = np.dot(state_prices, bonds)
stock_price_today = np.dot(stocks, state_prices)

# Find return expectations
bond_return = (np.dot(bonds, probabilities)/bond_price_today) - 1
stock_return = (np.dot(stocks, probabilities)/stock_price_today) - 1
print(f"Bond price today: {bond_price_today:.4f} and expected return : {(bond_return)*100:.4}%")
print(f"Stock price: ${stock_price_today:.4f} and expected return : {(stock_return)*100:.4f}%\n")

labels = ['bonds', 'stocks']
asset_returns = [bond_return, stock_return]

plt.figure(figsize=(8, 6))
plt.bar(labels, asset_returns, color=['#3498db', '#e67e22'])
plt.ylabel('expected return')
plt.title("Comparison of assets expected returns")
plt.axhline(0, color='black', linewidth=0.8)

for i, val in enumerate(asset_returns):
  numerical_val = float(val)
  plt.text(i, numerical_val+ 0.01, f'{numerical_val:.2%}', ha='center', fontweight='bold')
plt.show()
```

PV of inheritance

```python
# receiving inheritance today
# increase the interest rate with same maturities time
inherited = 1900000
discount_rate = 7/100
years_to_check = [0, 5, 10, 50]
years = np.arange(0, 51)

pv_values = inherited/(1 + discount_rate)**years
specific_pvs = [inherited/(1 + discount_rate)**t for t in years_to_check]

plt.figure(figsize=(10, 6))
plt.plot(years, pv_values, label="Present Value Curve", color='teal', linewidth=2)

plt.scatter(years_to_check, specific_pvs, color='red', zorder=5)
for i, txt in enumerate(years_to_check):
  plt.annotate(f"Year {txt}: ${specific_pvs[i]:,.0f}", (years_to_check[i], specific_pvs[i]), textcoords='offset points', xytext=(0,10), ha='center' )


plt.title("PV of of $1.9M Inheritance (3% Discount Rate)")
plt.xlabel("Years in the future")
plt.ylabel("Present Value ($)")
plt.gca().get_yaxis().set_major_formatter(plt.FuncFormatter(lambda x, p: format(int(), ',')))
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```
