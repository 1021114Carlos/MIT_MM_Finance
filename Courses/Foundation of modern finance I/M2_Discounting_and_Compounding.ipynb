#Libraries

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import math
from IPython.display import Math, display
from matplotlib.lines import lineStyles
```

Lecture code

```python
# Time and risk relation
t = np.linspace(0, 30, 100)    # time from 0 to 30 years
r_values = [0.04, 0.08, 0.12]
styles = [("black", "-"), ("red", "--"), ("blue", ":")]

# plotting
plt.figure(figsize=(10,6))

for r, (color, ls) in zip(r_values, styles):
  pv = 1/(1 + r)**t
  plt.plot(t, pv, label=f'r = {r}', color=color, linestyle=ls, linewidth=1.5)

plt.title('FV of $1 in year t', fontsize=14, pad=15)
plt.xlabel("Year when $1 is received", fontsize=14)
plt.ylabel('', fontsize=12)

# set axis limits
plt.xlim(0, 30)
plt.ylim(0, 1.1)

plt.xticks(np.arange(0, 31, 2))

plt.yticks(np.arange(0, 1.1, 0.2), [f'${val:.1f}' for val in np.arange(0, 1.1, 0.2)])

# add horizontal grid lines
plt.grid(axis='y', linestyle='-', alpha=0.7)
plt.grid(axis='x', linestyle='-', alpha=0.5)
plt.legend(loc='upper center', ncol=3, frameon=False, fontsize=11)

plt.tight_layout()
plt.show()
```

Calculate CF for business decision:

```python
def calculate_npv_with_details(name, cash_flows, initial_investment, rate):

  pvs = [cf / (1 + rate)**t for t, cf in enumerate(cash_flows, 1)]
  total_pv = sum(pvs)
  npv = total_pv - initial_investment

  print(f"--- {name} ---")
  for year, pv in enumerate(pvs, 1):
      print(f"Year {year} PV: {pv:,.4f}")

  print(f"Total PV: {total_pv:,.4f}")
  print(f"Net Present Value (NPV): {npv:,.4f}\n")

# Data setup
companies = {"Company A": {"flows": [500, 400, 300], "init": 1000},
             "Company B": {"flows": [-200, 300, 300], "init": 200}}

discount_rate = 0.05

# Run for all companies
for name, data in companies.items():
    calculate_npv_with_details(name, data["flows"], data["init"], discount_rate)
```
