Time & Risk relation

```python
from matplotlib.lines import lineStyles
pv = float(input("Enter amount to receive $: "))
rate = float(input("Enter rate (5%, 6%): "))/100
time = float(input("Enter years (e.g. 1, 3, 4): "))
present_value = pv/(1 + rate)**time
print(present_value)
# plotting
plt.plot(present_value, pv, 'ro', lineStyles='dashed')
plt.axis((0,6, 0,10))
plt.xlabel("Time")
plt.ylabel("Present Value")
plt.title("PV of Future CF")
plt.show()
```
