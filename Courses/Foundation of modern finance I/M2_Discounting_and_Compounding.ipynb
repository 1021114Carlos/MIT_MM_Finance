```python
def solve_state_prices(payoff_matrix, price_vector):
  """Solves for state prices Phi inthe equation:
  payoff_matrix @phi = price_vector
  payoff_matrix: matrix of shape (assets x states)
  price_vector: vector of asset prices
  """

  payoff_matrix = np.array(payoff_matrix, dtype=float)
  price_vector = np.array(price_vector, dtype=float)

  phi = np.linalg.solve(payoff_matrix, price_vector)
  return phi

payoff_matrix = [[105, 105], [90, 120]]

price_vector = [100, 100]

phi = solve_state_prices(payoff_matrix, price_vector)

print("state prices (phi_1, phi_2):")
print(phi)

print(f"phi_1 = {phi[0]:.4f}, and {phi[1]:.4f}\nThese must be positive (no-arbitrage condition).")
```

```python
def clean_input_list(prompt):
  """ Reads user input and converts numbers separated by commas or spaces into a float list."""

  raw = input(prompt)
  raw = raw.replace(",", " ")
  parts = raw.split()
  return np.array([float(x) for x in parts])

def q4_assets_xy():
  #state prices
  phi = clean_input_list("Enter state prices φ1 φ2 φ3 (e.g. 0.2, 0.55, 0.2): ")

  #probabilities
  equal = input("Are states equally likely? (y/n): ").strip().lower()
  if equal == "y":
    probs = np.array([1/3, 1/3, 1/3])
  else:
    probs = clean_input_list("Enter probabilities p1, p2, p3: ")

  # Asset x
  x = clean_input_list("\nEnter payoffs for asset x (e.g. 100 80 100): ")

  # asset y
  y = clean_input_list("\nEnter payoffs for asset y (e.g. 100 100 80): ")

  # pricing formula
  px = np.dot(phi, x)
  py = np.dot(phi, y)

  # expected payoffs
  ex = np.dot(probs, x)
  ey = np.dot(probs, y)

  # returns
  rx = (ex - px)/px
  ry = (ey - py)/py

  print(f"Asset X price: {px:.2f}, expected payoff: {ex:.2f}, expected return: {rx*100:.2f}%")
  print(f"Asset Y price: {py:.2f}, expected payoff: {ey:.2f}, expected return: {ry*100:.2f}%")

q4_assets_xy()
```

```python
def q5_pv_choice():
  amount_now = float(input("Amount you can get today (e.g. 10,000): "))
  future_amount1 = float(input("Future amount option 1 (e.g. 11500): "))
  t1 = float(input("Years until option 1 (e.g. 1): "))
  future_amount2 = float(input("Future amount option 2 (e.g. 18,000): "))
  t2 = float(input("Years until option 2 (e.g. 10): "))
  r = float(input("Annual interest rate in % (e.g 8): "))/100

  pv1 = future_amount1/(1 + r)**t1
  pv2 = future_amount2/(1 + r)**t2

  print(f"\nPV of getting {future_amount1} in {t1} years: {pv1:.4f}")
  print(f"PV of getting {future_amount2} in {t2} years: {pv2:.4f}")
  print(f"Amount today: {amount_now:.4f}")

  # compare each option with today amount
  print("Option today vs. option 1:",
        "Choose otion 1" if pv1 > amount_now else "Choose today")

  print("Option today vs. option 2:", "Choose option 2" if pv2 > amount_now else "choose today")

q5_pv_choice()

```

```python
years = int(input("Enter the number of years (e.g. 20): "))
r = float(input("Enter annual return as decimal (e.g 10.2% -> 10.2): "))/100
principal = float(input("X (use 1.0 to show 'multiples of X'): "))

c = display(Math(r"w_c(t) = x(1+r)^t"))
s = display(Math(r"w_s(t) = x(1 + tr)"))
i_on_i = display(Math(r"w_c(t) - w_s(t)]"))  # Interest on interest

compound_t = principal*(1 + r)**years
simple_t = principal*(1 + years*r)
interest_on_interest_t = compound_t - simple_t

print(f"Compound: = {compound_t:.6f}")
print(f"Simple: = {simple_t:.6f}")
print(f"Interert on interest: = {interest_on_interest_t:.6f}")

# year-by-year table
rows = []

for t in range(0, years + 1):
  simple_balance = principal*(1 + r*t)
  compound_balance = principal*((1 + r)**t)

  # Interest earned during year t (t-1 -> t)
  if t == 0:
    simple_interest_earned = 0
    compound_interest_earned = 0
  else:
    simple_interest_earned = principal*r
    compound_interest_earned = principal*r*((1 + r)**(t - 1))

  interest_on_interest = compound_balance - simple_balance

  rows.append({"Year": t, "Simple balance": simple_balance,
               "Compound balance": compound_balance,
               "Simple interest earned (that year)": simple_interest_earned,
               "Compound interest earned (that year)": compound_interest_earned,
               "Interest-on-interest (cumulative)": interest_on_interest,
               })

df = pd.DataFrame(rows)

# Display tables

display_df = df.copy()
for col in display_df.columns:
  if col != "Year":
    display_df[col] = display_df[col].round(6)

display(display_df)

# plot 1: simple vs compounding balances

plt.figure()
plt.plot(df["Year"], df["Simple balance"], label="Simple balance w_s(t) = x(1+rt)")
plt.plot(df["Year"], df["Compound balance"], label="Compound balance w_c(t) = x(1 + r)^t")
plt.xlabel("Year")
plt.ylabel("Wealth")
plt.title("Growth Curves: Simple vs Compounding")
plt.legend()
plt.show()

# plot 2: Interest-on-interest (Cumulative)
plt.figure()
plt.plot(df["Year"], df["Interest-on-interest (cumulative)"], label="w_c(t) - w_s(t)")
plt.xlabel("Year")
plt.ylabel("Extra wealth from compounding")
plt.title("Interest-on-interest (Cumulative): x[(1 + r)^t - (1 + rt)]")
plt.legend()
plt.show()

# plot 3: Annual interest earned (per year)
plt.figure()
plt.plot(df["Year"], df["Simple interest earned (that year)"], label="Simple: i_s(t)=rx")
plt.plot(df["Year"], df["Compound interest earned (that year)"], label="Compound: i_c(t)=rx(1+r)^(t-1)")
plt.xlabel("Year")
plt.ylabel("Interest earned during the year")
plt.title("Annual Interest Earned: Simple vs Compound")
plt.legend()
plt.show()



```

```python
def q7_rental_real_nominal():
  rental_rev_0 = float(input("Last year's rental revenue (e.g. 65,000): "))
  real_g_r = float(input("Real growth rate in % (e.g. 3): "))/100
  inflation = float(input("Inflation rate in % (e.g. 2.5): "))/100
  real_d_r = float(input("Real discount rate in % (e.g. 6): "))/100
  years = int(input("Number of years (e.g. 4): "))

  # Par I: One-year calculations (if years >= 1)
  if years >= 1:
    real_rev_1 = rental_rev_0*(1 + real_g_r)
    real_pv_1 = real_rev_1/(1 + real_d_r)

    nom_r_1 = rental_rev_0*(1 + real_g_r)*(1 + inflation)
    nom_pv_1 = nom_r_1/((1 + real_d_r)*(1 + inflation))

    print(f"Real revenue next year: {real_rev_1:.4f}")
    print(f"PV of real revenue: {real_pv_1:.4f}")
    print(f"Nominal revenue next year: {nom_r_1:.4f}")
    print(f"PV of nominal revenue: {nom_pv_1:.4f}")

  #Part II: Multi-year
  real_revenues = []
  nominal_revenues = []

  nom_g = (1 + real_g_r)*(1 + inflation) - 1
  nom_r = (1 + real_d_r)*(1 + inflation) - 1

  for t in range(1, years + 1):
    R_real_t = rental_rev_0*(1 + real_g_r)**t
    R_nom_t = rental_rev_0*(1 + nom_g)**t
    real_revenues.append(R_real_t)
    nominal_revenues.append(R_nom_t)

  PV_real_total = sum(real_revenues[t-1]/(1 + real_d_r)**t for t in range (1, years+1))
  PV_nom_total = sum(nominal_revenues[t-1]/(1 + nom_r)**t for t in range(1, years+1))

  print("Year | Real Revenue | Nominal Revenue")
  for t in range(1, years+1):
    print(f"{t:>4} | {real_revenues[t-1]:>12.4f} | {nominal_revenues[t-1]:>15.4f}")

  print(f"\nTotal PV of real revenues: {PV_real_total:.4f}")
  print(f"Total PV of nominal revenues: {PV_nom_total:.4f}")
  print(f"(They should be very closed if inputs are consistent.)")

q7_rental_real_nominal()




```
