```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import math
from IPython.display import Math, display
```

#Lecture

```python
def solve_state_prices(payoff_matrix, price_vector):
  """Solves for state prices Phi inthe equation:
  payoff_matrix @phi = price_vector
  payoff_matrix: matrix of shape (assets x states)
  price_vector: vector of asset prices
  """

  payoff_matrix = np.array(payoff_matrix, dtype=float)
  price_vector = np.array(price_vector, dtype=float)

  phi = np.linalg.solve(payoff_matrix, price_vector)
  return phi

payoff_matrix = [[105, 105], [90, 120]]

price_vector = [100, 100]

phi = solve_state_prices(payoff_matrix, price_vector)

print("state prices (phi_1, phi_2):")
print(phi)

print(f"phi_1 = {phi[0]:.4f}, and {phi[1]:.4f}\nThese must be positive (no-arbitrage condition).")
```
