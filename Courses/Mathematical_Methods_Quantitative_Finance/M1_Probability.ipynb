```python
from scipy.stats import binom
import matplotlib.pyplot as plt
import numpy as np
import math
```

Recitation Code

```python
n_list= [1, 2, 5, 10, 50, 100, 1000]
p = 0.1
k = []
for n in n_list:
  k = np.arange(0, n + 1)
  f= binom.pmf(k, n, p)

  plt.figure(figsize=(8, 4))
  plt.bar(k, f, tick_label=k)
  plt.xlabel("Number of Successes")
  plt.ylabel("Probability")
  plt.title(f"Binomila Distribution, p={p} n={n}")
plt.show()
```

```python
# Rescaled sequence
zmax = 5

for n in n_list:
  k = np.arange(0, n + 1)
  z = (k - n*p)/np.sqrt(n*p*(1-p))

  # Filter indices where |z| <= zmax
  mask = (np.abs(z) <= zmax)
  zfiltered = z[mask]
  ffiltered = binom.pmf(k[mask], n, p)

  plt.figure(figsize=(8, 4))
  plt.bar(zfiltered, ffiltered, width=0.1)
  plt.xlabel("Scaling variable z")
  plt.ylabel("Probability")
  plt.title(f"Rescaled Binomial Distribution p={p} n={n}")
  plt.show()
```
