```python
import math
from scipy.stats import norm
```

(Q1.A)

```python
rfr = 7.18/100
t_coupon = 3/12
t = 8/12
vol = 9.26/100
bond_price_S0 = 910
K = 900

pv_coupon = 35*math.exp(-rfr*t_coupon)
S0_star = 910 - pv_coupon
F0 = S0_star*math.exp(rfr*t)
print(f"Forward bond price {F0:.4f}")

d1 = (math.log(F0/K) + 0.5*(vol**2)*t)/(vol*math.sqrt(t))
d2 = d1 - vol*math.sqrt(t)


N_d1 = norm.cdf(d1)
N_d2 = norm.cdf(d2)
N_minus_d1 = norm.cdf(-d1)
N_minus_d2 = norm.cdf(-d2)


P0 = math.exp(-rfr*t)*(K*N_minus_d2 - F0*N_minus_d1)
print(f"Put option {P0:.4f}")
```
